```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingClassifier, GradientBoostingRegressor
from sklearn.metrics import (accuracy_score, f1_score, roc_auc_score, roc_curve,
                             confusion_matrix, mean_absolute_error, mean_squared_error)
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

FILE = "/home/hopelessnerd/Downloads/SEM7/IDC409/ds_first_git/data/cleaned_matches.csv"                 # <- change to your path
CHECKPOINTS = [6, 10, 15]        # overs
TRAIN_FRACTION = 0.8             # oldest 80% of matches train, newest 20% test

```

```python
df = pd.read_csv(FILE, low_memory=False)
df.columns = df.columns.str.strip()
df["date"] = pd.to_datetime(df["date"])
df["runs_total"] = pd.to_numeric(df["runs_total"], errors="coerce").fillna(0)
df["over"] = pd.to_numeric(df["over"], errors="coerce").fillna(0)      # 0 = first over
df["is_wicket"] = df["wicket_kind"].notna().astype(int)

# same team / ground written in different ways -> make identical
team_fix = {"Royal Challengers Bengaluru": "Royal Challengers Bangalore",
            "Rising Pune Supergiants": "Rising Pune Supergiant",
            "Delhi Daredevils": "Delhi Capitals",
            "Kings XI Punjab": "Punjab Kings"}
for c in ["batting_team", "match_won_by", "toss_winner"]:
    df[c] = df[c].astype("string").str.strip().replace(team_fix)
df["venue"] = (df["venue"].str.split(",").str[0].str.strip()
                 .replace({"Feroz Shah Kotla": "Arun Jaitley Stadium",
                           "M.Chinnaswamy Stadium": "M Chinnaswamy Stadium"}))

inn1 = df[df["innings"] == 1]
inn2 = df[df["innings"] == 2]               # innings 3+ (super overs) ignored
print("Loaded:", df.shape)
```

```python
# ================================================================ 2. ONE ROW PER MATCH
first = df.drop_duplicates("match_id")[["match_id", "date", "venue", "match_won_by",
                                        "toss_winner", "toss_decision"]]
m = (first
     .merge(inn1.groupby("match_id").agg(score_1st=("runs_total", "sum")), on="match_id")
     .merge(inn2.groupby("match_id").agg(score_2nd=("runs_total", "sum")), on="match_id")
     .merge(inn1.drop_duplicates("match_id")[["match_id", "batting_team"]]
               .rename(columns={"batting_team": "bat_first"}), on="match_id")
     .merge(inn2.drop_duplicates("match_id")[["match_id", "batting_team"]]
               .rename(columns={"batting_team": "chaser"}), on="match_id"))
m = m[m["match_won_by"].notna()]                                  # drop no-result matches
m = m.sort_values(["date", "match_id"]).reset_index(drop=True)    # chronological order
m["chase_win"] = (m["match_won_by"] == m["chaser"]).astype(int)   # target for chase models
m["bat_first_won"] = 1 - m["chase_win"]                           # target for pre-match model
```

```python

# ================================================================ 3. FEATURE ENGINEERING
# 3a. Venue average first-innings total (past matches only)
league_avg = m["score_1st"].expanding().mean().shift(1).fillna(160)
m["venue_avg"] = (m.groupby("venue")["score_1st"]
                    .transform(lambda s: s.shift(1).expanding().mean()).fillna(league_avg))

# 3b. Venue bat-first win rate (past matches only)
m["venue_bat_first_rate"] = (m.groupby("venue")["bat_first_won"]
                               .transform(lambda s: s.shift(1).expanding().mean()).fillna(0.5))

# 3c. Recent form = win rate in each team's last 5 matches (past matches only)
long = pd.concat([m[["match_id", "date", "bat_first", "match_won_by"]].rename(columns={"bat_first": "team"}),
                  m[["match_id", "date", "chaser", "match_won_by"]].rename(columns={"chaser": "team"})])
long = long.sort_values(["date", "match_id"])
long["won"] = (long["team"] == long["match_won_by"]).astype(int)
long["form"] = (long.groupby("team")["won"]
                    .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean()).fillna(0.5))
form = long.set_index(["match_id", "team"])["form"]
m["bat_form"] = form.reindex(pd.MultiIndex.from_arrays([m["match_id"], m["bat_first"]])).values
m["chase_form"] = form.reindex(pd.MultiIndex.from_arrays([m["match_id"], m["chaser"]])).values
m["form_diff"] = m["bat_form"] - m["chase_form"]

# 3d. Toss features
m["toss_to_bat_first_team"] = (m["toss_winner"] == m["bat_first"]).astype(int)
m["toss_decision_bat"] = (m["toss_decision"] == "bat").astype(int)

# 3e. Powerplay score (FIRST innings only)
m = m.join(inn1[inn1["over"] < 6].groupby("match_id")["runs_total"].sum().rename("powerplay_runs"),
           on="match_id")

# 3f. Match state at each checkpoint: 1st innings (bat1_*) and chase (runs_/wkts_/needed_/rrr_)
for n in CHECKPOINTS:
    a = inn1[inn1["over"] < n].groupby("match_id").agg(**{f"bat1_runs_{n}": ("runs_total", "sum"),
                                                          f"bat1_wkts_{n}": ("is_wicket", "sum")})
    b = inn2[inn2["over"] < n].groupby("match_id").agg(**{f"runs_{n}": ("runs_total", "sum"),
                                                          f"wkts_{n}": ("is_wicket", "sum")})
    m = m.join(a, on="match_id").join(b, on="match_id")
    m[f"needed_{n}"] = m["score_1st"] - m[f"runs_{n}"]
    m[f"rrr_{n}"] = m[f"needed_{n}"] / (20 - n)            # runs per over still needed
m = m.dropna().reset_index(drop=True)

cut = int(len(m) * TRAIN_FRACTION)
train, test = m.iloc[:cut], m.iloc[cut:]
print(f"Matches: {len(m)} | train: {len(train)} | test: {len(test)}")

# small helpers -----------------------------------------------------------
def make_clf():
    return GradientBoostingClassifier(n_estimators=150, learning_rate=0.03, max_depth=3, random_state=42)

def make_reg():
    return GradientBoostingRegressor(n_estimators=150, learning_rate=0.03, max_depth=3, random_state=42)

def clf_report(y_true, pred, prob):
    return accuracy_score(y_true, pred), f1_score(y_true, pred), roc_auc_score(y_true, prob)

def reg_report(y_true, pred):
    return mean_absolute_error(y_true, pred), np.sqrt(mean_squared_error(y_true, pred))
```

```python
# ================================================================ 4. PRE-MATCH WINNER MODEL
print("\n=== 4. PRE-MATCH: does the team batting first win? ===")
pre_feats = ["bat_form", "chase_form", "form_diff", "venue_avg",
             "venue_bat_first_rate", "toss_to_bat_first_team", "toss_decision_bat"]
pre_clf = make_clf().fit(train[pre_feats], train["bat_first_won"])
pre_pred = pre_clf.predict(test[pre_feats])
pre_prob = pre_clf.predict_proba(test[pre_feats])[:, 1]
pre_acc, pre_f1, pre_auc = clf_report(test["bat_first_won"], pre_pred, pre_prob)
baseline = max(test["bat_first_won"].mean(), 1 - test["bat_first_won"].mean())
print(f"accuracy={pre_acc:.3f}  F1={pre_f1:.3f}  AUC={pre_auc:.3f}  (majority baseline={baseline:.3f})")
print(confusion_matrix(test["bat_first_won"], pre_pred))
```
